### Aggregating SKU Statistics

Group the filtered MTS finished goods by SKU and calculate the minimum, maximum, mean, median, variance, and standard deviation of order quantity, along with average lead time.

```python

sku_summary = mts_data.groupby("sku_number").agg(
    min_quantity=("order_quantity", "min"),
    max_quantity=("order_quantity", "max"),
    avg_quantity=("order_quantity", "mean"),
    median_quantity=("order_quantity", "median"),
    variance_quantity=("order_quantity", "var"),
    std_quantity=("order_quantity", "std"),
    avg_lead_time=("lead_time", "mean")
)

sku_summary.head()
```

- The codes above were used for these specific purposes 
    - Groupby- was used to group the rows by SKU in order to calculate easier 
    - Min/Max order quant- this tells us the range of the values
    - Avg/median quant- Shows the average and median quantities 
    - Variance_quant- measures the demand by variability
    - Std_quant- measures the standard deviation of each sku
    - avg_lead time tells us the avergae lead time per sku

### Question 2

Calculate safety stock for each SKU using the standard deviation of order quantity, average lead time, and Z-scores corresponding to service levels of 75%, 90%, and 95%.

```python

z_75 = norm.ppf(0.75)
z_90 = norm.ppf(0.90)
z_95 = norm.ppf(0.95)

print(z_75, z_90, z_95)
```

- the code norm.ppf helps return the scores for each service level that is provided 
    - using the print function the scores are displayed below the code as those valuses
        - 0.6744897501960817 
        - 1.2815515655446004 
        - 1.644853626951472

```python

sku_summary["safety_stock_75"] = (
    z_75 * sku_summary["std_quantity"] *
    np.sqrt(sku_summary["avg_lead_time"])
)

sku_summary["safety_stock_90"] = (
    z_90 * sku_summary["std_quantity"] *
    np.sqrt(sku_summary["avg_lead_time"])
)

sku_summary["safety_stock_95"] = (
    z_95 * sku_summary["std_quantity"] *
    np.sqrt(sku_summary["avg_lead_time"])
)

sku_summary.head()

```

- In this code the saftey stock formula is used in order to actually calculate it 
    - SS = z_a *[STD]_D * SqrtL
    - z_a = z_95
    - [STD]_D = sku_summary["std_quantity"]
    - SqrtL = np.sqrt(sku_summar["avg_lead_time"])
- The function .head() is then used in order to view a summary of our data
    - If you scroll down you see our saftey stock values for 75,90, and 95 are shown
